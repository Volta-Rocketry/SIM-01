```python
import json
from rocketpy import Rocket, SolidMotor
```

```python
with open("parameters_AURORA_v04.json", "r", encoding="utf-8-sig") as f:
    rocket_data = json.load(f)

airframe = rocket_data["airframe"]
nosecone = rocket_data["nosecone"]
fins = rocket_data["fins"]
```

```python
rocket = Rocket(
    radius=airframe["radius"],
    mass=airframe["mass"],
    inertia=tuple(airframe["inertia"]),
    power_on_drag=0.465,
    power_off_drag=0.465,
    center_of_mass_without_motor=airframe["center_of_mass_without_motor"],
    coordinate_system_orientation=airframe["coordinate_system_orientation"]
)

rocket.add_nose(length=nosecone["length"], kind=nosecone["kind"], position=0)

rocket.add_trapezoidal_fins(
    n=fins["number"],
    root_chord=fins["root_chord"],
    tip_chord=fins["tip_chord"],
    span=fins["span"],
    position=fins["position"],
    sweep_length=fins["sweep_length"],
    cant_angle=fins["cant_angle"]
)
```

```python
# Valores leídos de OpenRocket
cg_true = 1.75
cp_true = 2.95
mass_true = 25.587

# Tolerancias
TOL_CG_CP = 0.02
TOL_MASS = 0.2

# Lo que RocketPy calculó
cg_rpy = rocket.center_of_mass(0)
cp_rpy = rocket.cp_position(0)
mass_rpy = rocket.mass

print(f"CG   → OR: {cg_true:.3f} m  |  RPy: {cg_rpy:.3f} m  |  diff: {abs(cg_true - cg_rpy):.4f} m  |  {'✓' if abs(cg_true - cg_rpy) <= TOL_CG_CP else '✗ FALLA'}")
print(f"CP   → OR: {cp_true:.3f} m  |  RPy: {cp_rpy:.3f} m  |  diff: {abs(cp_true - cp_rpy):.4f} m  |  {'✓' if abs(cp_true - cp_rpy) <= TOL_CG_CP else '✗ FALLA'}")
print(f"Masa → OR: {mass_true:.3f} kg |  RPy: {mass_rpy:.3f} kg |  diff: {abs(mass_true - mass_rpy):.4f} kg |  {'✓' if abs(mass_true - mass_rpy) <= TOL_MASS else '✗ FALLA'}")
```
